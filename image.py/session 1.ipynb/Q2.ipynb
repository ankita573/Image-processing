```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# -----------------------------------
# Read Image
# -----------------------------------

img = cv2.imread("lenaa.jpg")

if img is None:
    print("Image not found!")
    exit()

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("Original image size:", img.shape)


# -----------------------------------
# Nearest Neighbor - Manual
# -----------------------------------

def nearest_neighbor(img, scale):

    h, w, c = img.shape

    new_h = int(h * scale)
    new_w = int(w * scale)

    output = np.zeros(
        (new_h, new_w, c),
        dtype=img.dtype
    )

    for y in range(new_h):
        for x in range(new_w):

            src_y = min(int(y / scale), h - 1)
            src_x = min(int(x / scale), w - 1)

            output[y, x] = img[src_y, src_x]

    return output


# -----------------------------------
# Bilinear
# -----------------------------------

def bilinear_resize(img, scale):

    h, w = img.shape[:2]

    new_w = int(w * scale)
    new_h = int(h * scale)

    return cv2.resize(
        img,
        (new_w, new_h),
        interpolation=cv2.INTER_LINEAR
    )


# -----------------------------------
# Bicubic
# -----------------------------------

def bicubic_resize(img, scale):

    h, w = img.shape[:2]

    new_w = int(w * scale)
    new_h = int(h * scale)

    return cv2.resize(
        img,
        (new_w, new_h),
        interpolation=cv2.INTER_CUBIC
    )


# -----------------------------------
# Nearest Neighbor
# -----------------------------------

nn_2x = nearest_neighbor(img, 2)
nn_4x = nearest_neighbor(img, 4)

nn_half = nearest_neighbor(img, 0.5)
nn_quarter = nearest_neighbor(img, 0.25)


# -----------------------------------
# Bilinear
# -----------------------------------

bilinear_2x = bilinear_resize(img, 2)
bilinear_4x = bilinear_resize(img, 4)

bilinear_half = bilinear_resize(img, 0.5)
bilinear_quarter = bilinear_resize(img, 0.25)


# -----------------------------------
# Bicubic
# -----------------------------------

bicubic_2x = bicubic_resize(img, 2)
bicubic_4x = bicubic_resize(img, 4)

bicubic_half = bicubic_resize(img, 0.5)
bicubic_quarter = bicubic_resize(img, 0.25)


# -----------------------------------
# Display Results
# -----------------------------------

plt.figure(figsize=(15, 12))

images = [
    (img, "Original"),

    (nn_2x, "Nearest Neighbor - 2x"),
    (bilinear_2x, "Bilinear - 2x"),
    (bicubic_2x, "Bicubic - 2x"),

    (nn_4x, "Nearest Neighbor - 4x"),
    (bilinear_4x, "Bilinear - 4x"),
    (bicubic_4x, "Bicubic - 4x"),

    (nn_half, "Nearest Neighbor - 1/2"),
    (bilinear_half, "Bilinear - 1/2"),
    (bicubic_half, "Bicubic - 1/2"),

    (nn_quarter, "Nearest Neighbor - 1/4"),
    (bilinear_quarter, "Bilinear - 1/4"),
]

for i, (image, title) in enumerate(images):

    plt.subplot(3, 4, i + 1)

    plt.imshow(image)
    plt.title(title)
    plt.axis("off")


plt.tight_layout()
plt.show()
```
