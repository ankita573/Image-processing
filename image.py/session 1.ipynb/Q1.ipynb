```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image using OpenCV
image = cv2.imread("lenaa.jpg")

if image is None:
    print("❌ Image not found. Check the filename.")
else:
    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Display using Matplotlib
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.show()

    # Height and Width
    height, width = gray.shape

    # Number of pixels
    number_of_pixels = height * width

    # Datatype
    datatype = gray.dtype

    # Minimum pixel intensity
    minimum_intensity = np.min(gray)

    # Maximum pixel intensity
    maximum_intensity = np.max(gray)

    # Mean intensity
    mean_intensity = np.mean(gray)

    # Center pixel
    row = height // 2
    column = width // 2
    pixel_value = gray[row, column]

    # Display information
    print("----- IMAGE INFORMATION -----")
    print("Height:", height)
    print("Width:", width)
    print("Number of Pixels:", number_of_pixels)
    print("Datatype:", datatype)
    print("Minimum Pixel Intensity:", minimum_intensity)
    print("Maximum Pixel Intensity:", maximum_intensity)
    print("Mean Pixel Intensity:", mean_intensity)
    print("Pixel Coordinate:", (row, column))
    print("Pixel Value:", pixel_value)
```
