```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 
```

```python
img = cv2.imread("lenaa.jpg")

# Convert BGR to RGB for displaying with Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
rows, cols = img.shape[:2]

tx = 100
ty = 50

translation_matrix = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])

translated = cv2.warpAffine(
    img,
    translation_matrix,
    (cols, rows)
)
```

```python
scaled = cv2.resize(
    img,
    None,
    fx=1.5,
    fy=1.5,
    interpolation=cv2.INTER_LINEAR
)
```

```python
center = (cols // 2, rows // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    45,
    1.0
)

rotated = cv2.warpAffine(
    img,
    rotation_matrix,
    (cols, rows)
)
```

```python
horizontal_reflection = cv2.flip(img, 1)
```

```python
vertical_reflection = cv2.flip(img, 0)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(cv2.cvtColor(translated, cv2.COLOR_BGR2RGB))
plt.title("Translation")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(cv2.cvtColor(scaled, cv2.COLOR_BGR2RGB))
plt.title("Scaling")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotation")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(cv2.cvtColor(horizontal_reflection, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Reflection")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(cv2.cvtColor(vertical_reflection, cv2.COLOR_BGR2RGB))
plt.title("Vertical Reflection")
plt.axis("off")

plt.tight_layout()
plt.show()
```
